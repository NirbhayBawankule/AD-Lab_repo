```python
x = np.array(x_list).reshape(-1, 1)
y = np.array(y_list)
```

```python
lin_reg = LinearRegression()
lin_reg.fit(x, y)

y_pred_lin = lin_reg.predict(x)

print("Simple Linear Equation:")
print(f"y = {lin_reg.coef_[0]} * x + {lin_reg.intercept_}")
```

```python
plt.scatter(x, y)
plt.plot(x, y_pred_lin)
plt.title("Simple Linear Regression")
plt.show()
```

```python
print("Linear Regression Metrics")
print("MAE:", mean_absolute_error(y, y_pred_lin))
print("MSE:", mean_squared_error(y, y_pred_lin))
print("RMSE:", np.sqrt(mean_squared_error(y, y_pred_lin)))
print("R2 Score:", r2_score(y, y_pred_lin))
```

```python
for d in range(2, 8):
    print(f"\nPolynomial Regression Degree {d}")
    
    poly = PolynomialFeatures(degree=d)
    x_poly = poly.fit_transform(x)
    
    model = LinearRegression()
    model.fit(x_poly, y)
    
    y_pred_poly = model.predict(x_poly)
    
    print("Equation Coefficients:", model.coef_)
    
    # Visualization
    plt.scatter(x, y)
    plt.plot(x, y_pred_poly)
    plt.title(f"Polynomial Regression Degree {d}")
    plt.show()
    
    # Metrics
    print("MAE:", mean_absolute_error(y, y_pred_poly))
    print("MSE:", mean_squared_error(y, y_pred_poly))
    print("RMSE:", np.sqrt(mean_squared_error(y, y_pred_poly)))
    print("R2 Score:", r2_score(y, y_pred_poly))
```

```python
rf = RandomForestRegressor(n_estimators=100, random_state=42)
rf.fit(x, y)

y_pred_rf = rf.predict(x)

print("\nRandom Forest Model Details:")
print(rf)
```

```python
plt.scatter(x, y)
plt.plot(x, y_pred_rf)
plt.title("Random Forest Regression")
plt.show()
```

```python
print("Random Forest Metrics")
print("MAE:", mean_absolute_error(y, y_pred_rf))
print("MSE:", mean_squared_error(y, y_pred_rf))
print("RMSE:", np.sqrt(mean_squared_error(y, y_pred_rf)))
print("R2 Score:", r2_score(y, y_pred_rf))
```
